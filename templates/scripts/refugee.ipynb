# Planet Names

```python
import random
import roman



class celestialobjectname():

    def __init__(self, desc, moonORplanet, name, primarybody):
        self.desc = desc
        self.moonORplanet = moonORplanet
        self.name = name
        self.primarybody = primarybody

        # Convert inputs to strings and clean whitespace
        self.desc = str(self.desc).strip() if self.desc is not None else ""
        self.moonORplanet = str(self.moonORplanet).lower() if self.moonORplanet is not None else ""
        self.name = str(self.name) if self.name is not None else ""
        self.primarybody = str(self.primarybody) if self.primarybody is not None else ""

    def callknownname(self):

        # Part 1: Prefix
        prefix = f"{self.desc} " if self.desc else ""

        # Parts 2 & 3: Type and Suffix logic
        if "moon" in self.moonORplanet:
            obj_type = "Moon"
            suffix = f" of {self.primarybody}"
        elif "planet" in self.moonORplanet:
            obj_type = "Planet"
            suffix = f" {self.primarybody}"
        else:
            obj_type = "WHAT IS THIS"
            suffix = ""

        # Combine parts into a single string
        return f"{self.name}, {prefix}{obj_type}{suffix}"

    def callunknownname(self):
                # Generates a random integer where 1 <= number <= 10
        random_number = roman.toRoman(random.randint(1, 10))
        return f"{self.primarybody} {random_number}"

obj1 = celestialobjectname("Feline", "moon", "Abc", "NyaNya")
print(obj1.callknownname())
print(obj1.callunknownname())
```

# Ship Names

```python
import roman
from typing import Annotated
from pydantic import Field, validate_call

# 1. Define your reusable strict type rules
TwoToThreeLetters = Annotated[str, Field(pattern=r"^[A-Za-z]{2,3}$")]
AlphaNumericMix   = Annotated[str, Field(pattern=r"^[A-Za-z0-9]{2}$")]
SingleDigit       = Annotated[int, Field(ge=0, le=9)]
SingleAlphaNum    = Annotated[str, Field(pattern=r"^[A-Za-z0-9]$")]

class FormulaTranslator:
    
    # Automatically triggers runtime validation on all input arguments
    @validate_call
    def __init__(
        self, 
        architectural_code: TwoToThreeLetters,          # D
        model_code: AlphaNumericMix,                          # E
        model_changes: SingleDigit,                           # F
        equipment_updated: SingleDigit,                       # G
        model_name: str="",              # H
        prefix: str="",                  # J
        ship_subtype: str="",            # K
        adjective="",               # M
        ship_class: str="",              # N (renamed from 'class' as it is a reserved word)
        ship_classification2: str="",             # O
        owner: str="",                   # Q
        ship_name: str="",               # R
        affiliation: str=""              # T
        ):               
        """
        Accepts specific data attributes corresponding to the original Excel columns.
        All fields default to empty strings to prevent errors if attributes are omitted.
        """
        # Column mappings
        self.architectural_code = self._clean(architectural_code)
        self.model_code = self._clean(model_code)
        self.model_changes = self._clean(model_changes)
        self.equipment_updated = self._clean(equipment_updated)
        self.model_name = self._clean(model_name)
        self.prefix = self._clean(prefix)
        self.ship_subtype = self._clean(ship_subtype)
        self.adjective = self._clean(adjective)
        self.ship_class = self._clean(ship_class) 
        self.ship_class2 = self._clean(ship_classification2)
        self.owner = self._clean(owner)
        self.ship_name = self._clean(ship_name)
        self.affiliation = self._clean(affiliation)

    def _clean(self, val):
        """Helper to cast to string, strip whitespace, and handle None types."""
        return str(val).strip() if val is not None else ""

    def get_part1_r(self):
        return f'"{self.ship_name}", ' if self.ship_name else ""

    def get_part2_q(self):
        return f"{self.owner}'s " if self.owner else ""

    def get_part3_t(self):
        return f"{self.affiliation} " if self.affiliation else ""

    def get_part4_core(self):
        return f"{self.architectural_code}-{self.model_code}{self.model_changes}"

    def get_part5_6_ho(self):
        if self.model_name:
            return f" {self.model_name}-{self.ship_class2}" if self.ship_class2 else f" {self.model_name}"
        return self.ship_class2 if self.ship_class2 else ""

    def get_part7_g(self):
        if self.equipment_updated:
            try:
                return f" Mark-{roman.toRoman(int(float(self.equipment_updated)))}"
            except (ValueError, TypeError):
                pass
        return ""

    def get_part8_m(self):
        return f" {self.adjective}" if self.adjective else ""

    def get_part9_n(self):
        return f" {self.ship_class}-class" if self.ship_class else ""

    def get_part10_kj(self):
        if not self.ship_subtype:
            return ""
        if self.prefix:
            if " " in self.ship_subtype:
                space_idx = self.ship_subtype.find(" ")
                return f" {self.ship_subtype[:space_idx]} {self.prefix}-{self.ship_subtype[space_idx + 1:]}"
            return f" {self.prefix}-{self.ship_subtype}"
        return f" {self.ship_subtype}"

    def generate(self):
        """Concatenates all parts to generate the final string output."""
        return (
            self.get_part1_r() +
            self.get_part2_q() +
            self.get_part3_t() +
            self.get_part4_core() +
            self.get_part5_6_ho() +
            self.get_part7_g() +
            self.get_part8_m() +
            self.get_part9_n() +
            self.get_part10_kj()
        )
    
    def generate_model_string(self):
        """
        Processes the specific sub-section from Row 21:
        Core + HO Logic + Roman Mark + KJ Nested String
        """
        return (
            self.get_part4_core() +
            self.get_part5_6_ho() +
            self.get_part7_g() +
            self.get_part10_kj()
        )

# ==========================================
# Example Usage:
# ==========================================
if __name__ == "__main__":
    # You can now instantiate the class directly with named variables
    translator = FormulaTranslator(
        ship_name="Star Rupture",             # R
        owner="Bima",                   # Q
        affiliation="Rebellion",            # T
        architectural_code="SS",       # D
        model_code="20",               # E
        model_changes=9,              # F
        model_name="Starlancer",              # H
        ship_classification2="A",             # O
        equipment_updated=4,            # G
        adjective="Explorer",             # M
        ship_class="Luxury",             # N
        ship_subtype="Shuttle",          # K
        prefix="Light"                  # J
    )
    
    print(translator.generate())
    print(translator.generate_model_string())
```
